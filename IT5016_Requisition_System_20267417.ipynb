```python
# author: Joinda

"""
Create a class called RequisitionSystem.

Identify the shared data from your lab activity in Part A and use the __init__ method to initialize each of them (e.g., Date, Staff ID, Staff Name, status).
In the class RequsitionSystems, create methods for staff_info, requisitions_details, requisition_approval, respond_requsition, display_requisitons, requisition_statistic

Test your program by doing the following:
a. Create different instances of submitting a requisition. Submit at least 5 requisitions taking into account different scenarios. 
b. Display the requisition statistics.
c. Respond to the requisitions (approved or not approved or pending), then display the requisition information and statistics.
"""

# Global counter for unique requisition IDs
requisition_counter = 10000
Requisitions = []


class RequisitionSystem:
    # ============================
    # CONSTRUCTOR
    # ============================
    def __init__(self, date, staff_id, staff_name, items_with_costs):
        global requisition_counter
        requisition_counter += 1
        self.requisition_id = requisition_counter

        # Store staff info
        self.date = date
        self.staff_id = staff_id
        self.staff_name = staff_name

        # Store items and calculate total
        self.items_with_costs = items_with_costs
        self.total = sum(cost for _, cost in items_with_costs)

        # Initial status
        self.status = "Pending"
        self.approval_reference = "Not available"

        # Auto-approval rule
        if self.total < 500:
            self.status = "Approved"
            self.approval_reference = self.staff_id + str(self.requisition_id)[-3:]

    # ============================
    # UPDATE STAFF INFO
    # ============================
    def staff_info(self, date, staff_id, staff_name):
        self.date = date
        self.staff_id = staff_id
        self.staff_name = staff_name

    # ============================
    # UPDATE ITEMS + TOTAL
    # ============================
    def requisitions_details(self, items_with_costs):
        self.items_with_costs = items_with_costs
        self.total = sum(cost for _, cost in items_with_costs)
        return self.total

    # ============================
    # APPROVAL LOGIC
    # ============================
    def requisition_approval(self):
        self.status = "Pending"
        self.approval_reference = "Not available"

        if self.total < 500:
            self.status = "Approved"
            self.approval_reference = self.staff_id + str(self.requisition_id)[-3:]

    # ============================
    # UPDATE STATUS
    # ============================
    def update_status(self, new_status):
        new_status = new_status.strip().lower()

        if new_status == "approved":
            self.status = "Approved"
            self.approval_reference = self.staff_id + str(self.requisition_id)[-3:]
        elif new_status == "not approved" or new_status == "notapproved":
            self.status = "Not approved"
            self.approval_reference = "Not available"
        else:
            self.status = "Pending"
            self.approval_reference = "Not available"

    # ============================
    # MANAGER RESPONSE TO ONE REQUISITION
    # ============================
    def respond_requisition(self, decision):
        self.update_status(decision)

    # ============================
    # DISPLAY ALL REQUISITIONS
    # ============================
   
    def display_requisitions(req_list):
        print("\nPrinting Requisitions:\n")
        for req in req_list:
            print(f"Date: {req.date}")
            print(f"Requisition ID: {req.requisition_id}")
            print(f"Staff ID: {req.staff_id}")
            print(f"Staff Name: {req.staff_name}")
            print(f"Total: ${int(req.total)}")
            print(f"Status: {req.status}")
            print(f"Approval Reference Number: {req.approval_reference}")
            print()

    # ============================
    # STATISTICS
    # ============================
  
    def requisition_statistics(req_list):
        total_reqs = len(req_list)
        approved = sum(1 for r in req_list if r.status == "Approved")
        pending = sum(1 for r in req_list if r.status == "Pending")
        not_approved = sum(1 for r in req_list if r.status == "Not approved")

        print("\nStatistics:")
        print("Displaying the Requisition Statistics\n")
        print(f"The total number of requisitions submitted: {total_reqs}")
        print(f"The total number of approved requisitions: {approved}")
        print(f"The total number of pending requisitions: {pending}")
        print(f"The total number of not approved requisitions: {not_approved}")
        print()

    # ============================
    # MANAGER APPROVES ALL PENDING
    # ============================
    
    def manager_approve_pending(req_list):
        print("\nManager Approval Required for Pending Requisitions:\n")

        pending_found = False

        for req in req_list:
            if req.status == "Pending":
                pending_found = True

                print(f"Requisition ID: {req.requisition_id}")
                print(f"Date: {req.date}")
                print(f"Staff ID: {req.staff_id}")
                print(f"Staff Name: {req.staff_name}")
                print(f"Total: ${int(req.total)}")
                print(f"Current Status: {req.status}")

                decision = input("Enter decision (Approved / Not approved / Pending): ")
                req.update_status(decision)
                print("Status updated.\n")

        if not pending_found:
            print("No pending requisitions.\n")


# ============================
# MAIN PROGRAM
# ============================

r1 = RequisitionSystem("03/04/2026","FN19","Ammy Paul",[("Coffee", 100), ("Printer", 350)])
Requisitions.append(r1)

r2 = RequisitionSystem("05/09/2024", "FN20", "Tracy Brown", [("Laptop", 600), ("Bag", 400)])
Requisitions.append(r2)

r3 = RequisitionSystem("07/05/2024", "FN15", "Saad Abdullah", [("Server", 2000), ("Router", 1500)])
Requisitions.append(r3)

r4 = RequisitionSystem("03/05/2024", "FN02", "Asad Khan", [("Stationery", 490)])
Requisitions.append(r4)

r5 = RequisitionSystem("12/04/2025","IT12","Sam Smith",[("Coffee", 50), ("Printer", 350)])
Requisitions.append(r5)

# Display initial statistics
RequisitionSystem.requisition_statistics(Requisitions)

# Manager approves all pending requisitions
RequisitionSystem.manager_approve_pending(Requisitions)

# Display again after update
RequisitionSystem.display_requisitions(Requisitions)
RequisitionSystem.requisition_statistics(Requisitions)
```
